```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline 
```

```python
df=pd.read_csv("titanic.csv")
```

```python
df.head()
```

```python
# split into target and input columns
inputs=df.drop(['PassengerId','Survived','Name','SibSp','Parch','Ticket','Cabin','Embarked'],axis=1)
```

```python
# this is our target
target=df['Survived']
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
sex_n=LabelEncoder()
```

```python
inputs['Sex_n']=sex_n.fit_transform(inputs['Sex'])
```

```python
# this is our inputs
new_inputs=inputs.drop(['sex_n','Sex'],axis=1)
```

```python
from sklearn import tree
```

```python
model=tree.DecisionTreeClassifier()
```

```python
model.fit(new_inputs,target)
```

```python
model.score(new_inputs,target)
```
