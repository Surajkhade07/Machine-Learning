```python
import pandas as pd
from sklearn.datasets import load_digits 
```

```python
digits=load_digits()
```

```python
digits.keys()
```

```python
digits.data[0].reshape(8,8)
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
plt.gray()
plt.matshow(digits.data[0].reshape(8,8))
plt.show()
```

```python
df=pd.DataFrame(digits.data)
```

```python
df.head()
```

```python
X= df
Y=digits.target
```

```python
# scale th data
from sklearn.preprocessing import StandardScaler
scale=StandardScaler()
```

```python
X_scale=scale.fit_transform(X)
```

```python
# now we split data into train and test dataset
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(X_scale,Y,test_size=0.2,random_state=30)
```

```python
# use linear regression to train our data
from sklearn.linear_model import LogisticRegression
```

```python
model=LogisticRegression()
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

```python
# to reduce size od data remove less use dataset we use PCA 
from sklearn.decomposition import PCA
```

```python
pca=PCA(0.95)
x_pca=pca.fit_transform(X)
x_pca.shape
```

```python
# here it remove less use columns and from 64 it takesdown to 29
```

```python
x_train_pca,x_test_pca,y_train_pca,y_test_pca=train_test_split(x_pca,Y,test_size=0.2,random_state=30)
```

```python
model=LogisticRegression(max_iter=1000)
model.fit(x_train_pca,y_train_pca)
```

```python
model.score(x_test_pca,y_test_pca)
```
