```python
import pandas as pd
```

```python
df=pd.read_csv("heart.csv")
```

```python
df.head()
```

```python
# convert text column into numeric value
from sklearn.preprocessing import OrdinalEncoder
le=OrdinalEncoder()
```

```python
df[['ChestPainType','Sex','RestingECG','ExerciseAngina','ST_Slope']]=le.fit_transform(df[['ChestPainType','Sex','RestingECG','ExerciseAngina','ST_Slope']])
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x=df.drop('HeartDisease',axis=1)
y=df.HeartDisease
x.shape
```

```python
from sklearn.preprocessing import StandardScaler
scale=StandardScaler()
x_scale=scale.fit_transform(x)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x_scale,y,test_size=0.2,random_state=30)
```

```python
# create function that checkthe score of different model
def checkscore(model,x_train,x_test,y_train,y_test):
    model.fit(x_train,y_train)
    print(model.score(x_test,y_test))
```

```python
from sklearn.linear_model import LogisticRegression
checkscore(LogisticRegression(max_iter=1000),x_train,x_test,y_train,y_test)
```

```python
from sklearn.tree import DecisionTreeClassifier
checkscore(DecisionTreeClassifier(),x_train,x_test,y_train,y_test)
```

```python
from sklearn.svm import SVC
checkscore(SVC(),x_train,x_test,y_train,y_test)
```

```python
# to remove less use dataset 
from sklearn.decomposition import PCA
pca=PCA(0.95)
```

```python
x_pca=pca.fit_transform(x)
```

```python
x_train_pca,x_test_pca,y_train_pca,y_test_pca=train_test_split(x_pca,y,test_size=0.2,random_state=30)
```

```python
checkscore(LogisticRegression(),x_train_pca,x_test_pca,y_train_pca,y_test_pca)
```
